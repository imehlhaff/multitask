```python
# Item clusters: Spearman correlations between the 10 survey items on the train split, clustered with
# average linkage on 1 - correlation. Reference set excluded: 70 comments make up 32% of rows.

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.patches import Patch
from scipy.cluster.hierarchy import dendrogram, fcluster, linkage
from scipy.spatial.distance import squareform

ITEMS = ["sentiment", "respect", "insult", "humiliate", "status", "dehumanize", "violence", "genocide", "attack_defend", "hatespeech"]

train = pd.read_csv(Path.cwd() / "clean" / "hatespeech_train.csv", usecols=ITEMS + ["is_reference"])
corr = train.loc[~train["is_reference"], ITEMS].corr("spearman")

Z = linkage(squareform(1 - corr.values, checks=False), "average")
clusters = dict(zip(ITEMS, fcluster(Z, 3, "maxclust")))

# Key each cluster by a member item so colors follow the cluster, not dendrogram order
CLUSTER_STYLE = {
    clusters["respect"]: ("Hostility (+ dehumanize)", "#2a78d6"),
    clusters["violence"]: ("Violence", "#eb6834"),
    clusters["hatespeech"]: ("Hate speech judgement", "#1baf7a"),
}
MERGE_COLOR = "#a3a29c"


def leaves_under(link_id):
    if link_id < len(ITEMS):
        return [link_id]
    left, right = Z[link_id - len(ITEMS), :2].astype(int)
    return leaves_under(left) + leaves_under(right)


def link_color(link_id):
    # A link takes its cluster's color only if every item under it is in that cluster
    members = {clusters[ITEMS[i]] for i in leaves_under(link_id)}
    return CLUSTER_STYLE[members.pop()][1] if len(members) == 1 else MERGE_COLOR


fig, (ax_tree, ax_heat) = plt.subplots(1, 2, figsize=(11, 5.5), gridspec_kw={"width_ratios": [1, 2.2], "wspace": 0.02})

# Dendrogram on the left; leaves run top to bottom in the same order as the heatmap rows
tree = dendrogram(Z, orientation="left", labels=ITEMS, link_color_func=link_color, ax=ax_tree)
# Leaf dots carry the cluster color, including the single-item hatespeech cluster that has no branch to color
leaf_y = 5 + 10 * np.arange(len(ITEMS))
ax_tree.scatter(np.zeros(len(ITEMS)), leaf_y, s=60, c=[CLUSTER_STYLE[clusters[i]][1] for i in tree["ivl"]],
                edgecolors="white", linewidths=2, zorder=3, clip_on=False)
ax_tree.invert_yaxis()
ax_tree.set_yticks([])
ax_tree.set_xlabel("1 − Spearman ρ  (merge distance)")
for side in ("top", "right", "left"):
    ax_tree.spines[side].set_visible(False)

# Heatmap in leaf order on a one-hue ramp: light = weak, dark = strong
order = tree["ivl"]
cmap = LinearSegmentedColormap.from_list("blue", ["#f4f8fe", "#9ec5f4", "#3987e5", "#1c5cab", "#0d366b"])
masked = corr.loc[order, order].mask(np.eye(len(order), dtype=bool))
im = ax_heat.imshow(masked, cmap=cmap, vmin=0, vmax=1)
ax_heat.set_xticks(range(len(order)), order, rotation=45, ha="right")
ax_heat.set_yticks(range(len(order)), order)
ax_heat.yaxis.tick_right()
ax_heat.tick_params(length=0)
for i in range(len(order)):
    for j in range(len(order)):
        r = masked.iat[i, j]
        if not np.isnan(r):
            ax_heat.text(j, i, f"{r:.2f}", ha="center", va="center", fontsize=8, color="white" if r > 0.55 else "#0b0b0b")
for spine in ax_heat.spines.values():
    spine.set_visible(False)
fig.colorbar(im, ax=ax_heat, fraction=0.04, pad=0.2, label="Spearman ρ")

fig.legend(
    handles=[Patch(color=c, label=name) for name, c in CLUSTER_STYLE.values()],
    loc="upper center", ncol=3, frameon=False, bbox_to_anchor=(0.5, 1.0),
)
fig.suptitle("Survey items split into hostility, violence and the overall judgement", y=1.06)
plt.show()
```
