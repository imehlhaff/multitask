```python
from pathlib import Path

import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 200)

RAW_DIR = Path.cwd() / "raw"
CLEAN_PATH = Path.cwd() / "clean" / "cap_manifestos.csv"

# Rows to show per dataset. Change SEED to see different rows.
N = 5
SEED = 0
```

## Raw

```python
raw_uk = pd.read_csv(RAW_DIR / "uk_manifestos_v1__1.csv", encoding="cp1252", low_memory=False)
raw_irl = pd.read_csv(RAW_DIR / "irl_manifestos_data_1981_2020_utf8.csv", low_memory=False)
raw_us_dem = pd.read_csv(RAW_DIR / "Dem_partyplatform_20.2.csv", low_memory=False)
raw_us_rep = pd.read_csv(RAW_DIR / "US-Party-Rep_Platform_19.3_0.csv", low_memory=False)
```

### UK manifestos

```python
raw_uk.sample(N, random_state=SEED)
```

### Irish manifestos

```python
raw_irl.sample(N, random_state=SEED)
```

### US Democratic platforms

```python
raw_us_dem.sample(N, random_state=SEED)
```

### US Republican platforms

```python
raw_us_rep.sample(N, random_state=SEED)
```

## Clean

```python
clean = pd.read_csv(CLEAN_PATH, low_memory=False)
clean.groupby("country").sample(N, random_state=SEED)
```
